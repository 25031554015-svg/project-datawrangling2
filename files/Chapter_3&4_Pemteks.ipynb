```python
from apify_client import ApifyClient

client = ApifyClient("apify_api_BT3D5s0nXRVORI2WowAoem5IIumC9C3xRQ5S")

print("✅ Apify client siap")
```

```python
post_url = "https://www.instagram.com/reel/DUZPjdnk1Xw/?stkn=MXkyNW4zcWptc3lrOQ=="

jumlah_komentar = 300

print("URL:", post_url)
print("Target komentar:", jumlah_komentar)
```

```python
run = client.actor("apify/instagram-comment-scraper").call(
    run_input={
        "directUrls": [post_url],
        "resultsLimit": jumlah_komentar
    }
)

print("✅ Scraping selesai")
print("Dataset ID:", run.default_dataset_id)
```

```python
comments = list(
    client.dataset(run.default_dataset_id).iterate_items()
)

print("Jumlah komentar:", len(comments))
```

```python
for i, comment in enumerate(comments, 1):
    print(f"{i}.", comment)
```

```python
run = client.actor("apify/instagram-comment-scraper").call(
    run_input={
        "directUrls": [post_url],
        "resultsLimit": 300,
        "includeNestedComments": True
    }
)

print("Dataset ID:", run.default_dataset_id)
```

```python
comments = list(
    client.dataset(run.default_dataset_id).iterate_items()
)

print("Jumlah komentar:", len(comments))
```

```python
import pandas as pd

df = pd.DataFrame(comments)

df.to_excel("komentar_instagram.xlsx", index=False)

print(f"✅ Excel dibuat: {len(df)} komentar")
```
